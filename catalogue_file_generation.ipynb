```python
import pandas as pd
from faker import Faker
import random

fake = Faker()

categories = ['Accounting Software', 'HR Software', 'CRM Software',
              'ERP Software', 'Project Management', 'Cybersecurity',
              'Collaboration Tools', 'Inventory Management']

brands = ['SAP', 'Zoho', 'Tally', 'Salesforce', 'Freshworks',
          'Microsoft', 'Oracle', 'HubSpot', 'Quickbooks', 'Bitdefender']

stock_status = ['Active', 'Inactive', 'Discontinued']

def random_description():
    choices = [
        fake.sentence(nb_words=12),
        fake.sentence(nb_words=4),   # intentionally short = bad quality
        "",                           # intentionally blank = missing
        fake.paragraph(nb_sentences=2)
    ]
    return random.choices(choices, weights=[50, 20, 15, 15])[0]

def random_image_url():
    return random.choices(
        [f"https://images.example.com/{fake.uuid4()}.jpg", "", None],
        weights=[70, 20, 10]
    )[0]

rows = []
sku_pool = [f"SKU{str(i).zfill(4)}" for i in range(1, 101)]

# Inject ~8 duplicate SKUs
sku_list = sku_pool.copy()
for _ in range(8):
    sku_list.append(random.choice(sku_pool[:50]))

random.shuffle(sku_list)

for sku in sku_list:
    rows.append({
        "SKU_ID": sku,
        "Product_Name": random.choices(
            [fake.bs().title(), ""],
            weights=[90, 10]
        )[0],
        "Category": random.choices(
            categories + [""],
            weights=[10,10,10,10,10,10,10,10, 20]
        )[0],
        "Brand": random.choices(
            brands + [None],
            weights=[8,8,8,8,8,8,8,8,8,8, 20]
        )[0],
        "Description": random_description(),
        "Price": random.choices(
            [round(random.uniform(999, 99999), 2), None],
            weights=[80, 20]
        )[0],
        "Image_URL": random_image_url(),
        "Stock_Status": random.choices(
            stock_status + [None],
            weights=[60, 25, 10, 5]
        )[0],
        "Keywords": random.choices(
            [", ".join(fake.words(nb=4)), ""],
            weights=[75, 25]
        )[0],
    })

df = pd.DataFrame(rows)
df.to_csv("catalogue_raw.csv", index=False)
print(f"Dataset generated: {len(df)} rows")
print(df.isnull().sum())
```
